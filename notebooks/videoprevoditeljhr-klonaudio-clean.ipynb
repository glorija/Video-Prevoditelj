```python
import wave
import os

SOURCE = "/kaggle/input/datasets/zeljkozic/test-wav/source_voice.wav"
OUT = "/kaggle/working/source_voice_last60s.wav"

with wave.open(SOURCE, "rb") as src:
    rate = src.getframerate()
    channels = src.getnchannels()
    width = src.getsampwidth()
    total_frames = src.getnframes()

    frames_60s = int(60 * rate)

    src.setpos(total_frames - frames_60s)
    audio = src.readframes(frames_60s)

with wave.open(OUT, "wb") as dst:
    dst.setnchannels(channels)
    dst.setsampwidth(width)
    dst.setframerate(rate)
    dst.writeframes(audio)

print("=== ZADNJIH 60 SEKUNDI ===")
print("Izvor:", SOURCE)
print("Izlaz:", OUT)
print("Postoji:", os.path.exists(OUT))
print("Veličina:", os.path.getsize(OUT), "bytes")

with wave.open(OUT, "rb") as w:
    print("Kanali:", w.getnchannels())
    print("Sample rate:", w.getframerate(), "Hz")
    print("Sampleovi:", w.getnframes())
    print("Trajanje:", round(w.getnframes() / w.getframerate(), 3), "s")
```

```python
import subprocess
import sys

code = r'''
import torch
from kugelaudio_open import (
    KugelAudioForConditionalGenerationInference,
    KugelAudioProcessor,
)

MODEL = "RolandJAAI/klonaudio"
VOICE = "/kaggle/working/source_voice_last60s.wav"
OUT = "/kaggle/working/REFERENCE_01.wav"

TEXT = """A kako je do toga došlo? Dogodilo se tako da su ljudi zaboravili tko su zapravo.
A sustav je preuzeo vlast. Ljudi često ne primijete trenutak kada se stvari počnu mijenjati.
Tek kasnije shvate koliko se toga promijenilo i koliko je teško vratiti ono što je nekada bilo normalno."""

print("=== KLONAUDIO REFERENCE 01 ===")
print("Model:", MODEL)
print("Glas:", VOICE)
print("Izlaz:", OUT)
print()

print("Učitavam model...")
model = KugelAudioForConditionalGenerationInference.from_pretrained(
    MODEL,
    dtype=torch.bfloat16,
    low_cpu_mem_usage=True,
)

print("Učitavam processor...")
processor = KugelAudioProcessor.from_pretrained(MODEL)

print("Pripremam voice prompt...")
inputs = processor(
    text=TEXT,
    voice_prompt=VOICE,
    return_tensors="pt",
)

print("Generiram govor...")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        cfg_scale=3.0,
        max_new_tokens=2048,
    )

audio = outputs.speech_outputs[0]

print("Spremam WAV...")
processor.audio_processor.save_audio(audio, OUT)

print()
print("=== GOTOVO ===")
print("Izlaz:", OUT)
'''

result = subprocess.run(
    [sys.executable, "-c", code],
    capture_output=True,
    text=True
)

print(result.stdout)

print("=== STDERR ===")
print(result.stderr)

print("=== RETURN CODE ===")
print(result.returncode)
```

```python
import os
import wave
import numpy as np

PATH = "/kaggle/input/datasets/zeljkozic/test-wav/klonaudio_hr_60s_final_test.wav"

print("=== ANALIZA KLONAUDIO FORMATA ===")

with wave.open(PATH, "rb") as w:
    channels = w.getnchannels()
    sr = w.getframerate()
    sampwidth = w.getsampwidth()
    frames = w.getnframes()

print("Kanali:", channels)
print("Sample rate:", sr)
print("Sample width:", sampwidth)
print("Trajanje:", frames/sr)

with wave.open(PATH, "rb") as w:
    raw = w.readframes(frames)

if sampwidth == 2:
    audio = np.frombuffer(raw, dtype=np.int16).astype(np.float32)
elif sampwidth == 4:
    audio = np.frombuffer(raw, dtype=np.float32)
else:
    raise Exception("Nepoznat format")

audio = audio / (np.max(np.abs(audio)) + 1e-9)

print("Max:", np.max(np.abs(audio)))
print("RMS:", np.sqrt(np.mean(audio**2)))

# detekcija govora
energy = np.abs(audio)

threshold = np.percentile(energy, 35)

active = energy > threshold

changes = np.diff(active.astype(int))

starts = np.where(changes == 1)[0]
ends = np.where(changes == -1)[0]

if len(starts) and len(ends) and ends[0] < starts[0]:
    ends = ends[1:]

if len(starts) > len(ends):
    ends = np.append(ends, len(active)-1)

print("\n=== SEGMENTI ===")

segments=[]

for s,e in zip(starts,ends):
    dur=(e-s)/sr
    if dur > 0.15:
        segments.append((s/sr,e/sr,dur))

for i,(s,e,d) in enumerate(segments,1):
    print(f"{i:02}: {s:.3f} - {e:.3f} ({d:.3f}s)")

print("\nBroj segmenata:",len(segments))
print("Govor ukupno:",round(sum(x[2] for x in segments),3),"s")
```

```python
import wave
import numpy as np

PATH = "/kaggle/input/datasets/zeljkozic/test-wav/klonaudio_hr_60s_final_test.wav"

with wave.open(PATH, "rb") as w:
    sr = w.getframerate()
    audio = np.frombuffer(w.readframes(w.getnframes()), dtype=np.int16).astype(np.float32)

audio /= np.max(np.abs(audio))

# blokovi 50 ms
block = int(sr * 0.05)

rms = []

for i in range(0, len(audio), block):
    part = audio[i:i+block]
    rms.append(np.sqrt(np.mean(part**2)))

rms = np.array(rms)

print("=== RMS ANALIZA ===")
print("Min:", rms.min())
print("Max:", rms.max())
print("Median:", np.median(rms))

# ispis svakih 0.5 sekundi
print("\n=== ENERGIJA PO 0.5s ===")

step = int(0.5/0.05)

for i in range(0,len(rms),step):
    t=i*0.05
    print(f"{t:6.2f}s  {rms[i]:.4f}")
```

```python
import wave
import numpy as np

PATH = "/kaggle/input/datasets/zeljkozic/test-wav/klonaudio_hr_60s_final_test.wav"

with wave.open(PATH, "rb") as w:
    sr = w.getframerate()
    audio = np.frombuffer(
        w.readframes(w.getnframes()),
        dtype=np.int16
    ).astype(np.float32)

audio /= np.max(np.abs(audio))

# analiza u blokovima 20 ms
block = int(sr * 0.02)

energy = []

for i in range(0, len(audio), block):
    part = audio[i:i+block]
    energy.append(np.sqrt(np.mean(part**2)))

energy = np.array(energy)

# prag za tišinu
threshold = 0.015

silence = energy < threshold

# spoji tišine
changes = np.diff(silence.astype(int))

starts = np.where(changes == 1)[0]
ends = np.where(changes == -1)[0]

if len(ends) and len(starts) and ends[0] < starts[0]:
    ends = ends[1:]

if len(starts) > len(ends):
    ends = np.append(ends, len(silence)-1)

print("=== PAUZE >= 100 ms ===")

pauze = []

for s,e in zip(starts,ends):
    start = s * 0.02
    end = e * 0.02
    dur = end-start
    
    if dur >= 0.1:
        pauze.append((start,end,dur))

for i,(s,e,d) in enumerate(pauze,1):
    print(f"{i:02}: {s:7.3f} - {e:7.3f}  ({d:5.3f}s)")

print("\nUkupno pauza:",len(pauze))
print("Ukupno trajanje pauza:",
      round(sum(x[2] for x in pauze),3),"s")
```

```python
from pydub import AudioSegment

INPUT = "/kaggle/input/datasets/zeljkozic/test-wav/klonaudio_hr_60s_final_test.wav"
OUTPUT = "/kaggle/working/klonaudio_hr_60s_natural_pause.wav"

audio = AudioSegment.from_wav(INPUT)

# dodatne pauze u milisekundama
# vrijeme u sekundama -> mjesto ubacivanja
pauses = [
    (2.32, 1060),
    (4.68, 1440),
    (8.26, 960),
    (10.94, 500),
    (12.12, 620),
    (14.72, 440),
    (21.34, 1120),
    (23.16, 800),
    (25.20, 980),
    (29.10, 620),
    (32.26, 520),
    (40.76, 800),
]

# radimo od kraja prema početku
for position, extra in sorted(pauses, reverse=True):
    ms = int(position * 1000)
    silence = AudioSegment.silent(duration=extra)
    audio = audio[:ms] + silence + audio[ms:]

# završna pauza
audio += AudioSegment.silent(duration=5000)

audio.export(OUTPUT, format="wav")

print("=== GOTOVO ===")
print(OUTPUT)
print("Trajanje:", round(len(audio)/1000,3), "s")
```

```python
import os
import glob
import json
import subprocess
from collections import defaultdict

print("=== PRIPREMA VIDEA ZA DIARIZACIJU ===")

# Pronađi naš trenutni video
videos = glob.glob(
    "/kaggle/input/models/zeljkozic/ljudi-igraju-u-zivot-a-ko-igra-ljudima-shorts-mp4/other/default/1/*.mp4"
)

if not videos:
    raise FileNotFoundError("Video nije pronađen.")

video_path = videos[0]
audio_path = "/kaggle/working/current_video_audio.wav"

print("Video:", video_path)

# Izvuci mono 16 kHz WAV
subprocess.run([
    "ffmpeg", "-y",
    "-i", video_path,
    "-vn",
    "-ac", "1",
    "-ar", "16000",
    "-c:a", "pcm_s16le",
    audio_path
], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

print("Audio:", audio_path)

print("\n=== DIARIZACIJA ===")

result = pipeline(audio_path)

# Community-1 može vratiti objekt s poljem speaker_diarization
if hasattr(result, "speaker_diarization"):
    diarization = result.speaker_diarization
else:
    diarization = result

segments = []
speaker_totals = defaultdict(float)

for turn, _, speaker in diarization.itertracks(yield_label=True):
    start = float(turn.start)
    end = float(turn.end)
    duration = end - start

    segments.append({
        "start": start,
        "end": end,
        "duration": duration,
        "speaker": str(speaker)
    })

    speaker_totals[str(speaker)] += duration

# Spremi rezultat
out_json = "/kaggle/working/diarization_segments.json"

with open(out_json, "w", encoding="utf-8") as f:
    json.dump(segments, f, ensure_ascii=False, indent=2)

print("\n=== REZULTAT ===")
print("Ukupno segmenata:", len(segments))
print("Broj govornika:", len(speaker_totals))

for speaker, total in sorted(speaker_totals.items()):
    print(f"{speaker}: {total:.3f} s")

print("\nSpremljeno:", out_json)
```

```python
import json
import wave
import numpy as np
from collections import defaultdict

print("=== IZDVAJANJE GLASOVA GOVORNIKA ===")

audio_path = "/kaggle/working/current_video_audio.wav"
json_path = "/kaggle/working/diarization_segments.json"

with open(json_path, "r", encoding="utf-8") as f:
    segments = json.load(f)

# Učitaj PCM WAV
with wave.open(audio_path, "rb") as wf:
    sample_rate = wf.getframerate()
    channels = wf.getnchannels()
    sample_width = wf.getsampwidth()
    nframes = wf.getnframes()
    raw = wf.readframes(nframes)

if channels != 1 or sample_width != 2:
    raise RuntimeError(
        f"Očekivan mono 16-bit WAV, dobiveno: channels={channels}, sample_width={sample_width}"
    )

audio = np.frombuffer(raw, dtype=np.int16)

print("Sample rate:", sample_rate)
print("Trajanje:", round(len(audio) / sample_rate, 3), "s")

# Grupiraj segmente po govorniku
speaker_segments = defaultdict(list)

for seg in segments:
    speaker_segments[seg["speaker"]].append(seg)

output_files = []

for speaker in sorted(speaker_segments):
    pieces = []

    for seg in speaker_segments[speaker]:
        start = max(0, int(round(seg["start"] * sample_rate)))
        end = min(len(audio), int(round(seg["end"] * sample_rate)))

        if end > start:
            pieces.append(audio[start:end])

    if not pieces:
        continue

    # Mala prirodna pauza između izdvojenih govornih segmenata
    pause = np.zeros(int(0.10 * sample_rate), dtype=np.int16)

    joined = []
    for i, piece in enumerate(pieces):
        joined.append(piece)
        if i < len(pieces) - 1:
            joined.append(pause)

    speaker_audio = np.concatenate(joined)

    out_path = f"/kaggle/working/{speaker}_voice_from_video.wav"

    with wave.open(out_path, "wb") as wf:
        wf.setnchannels(1)
        wf.setsampwidth(2)
        wf.setframerate(sample_rate)
        wf.writeframes(speaker_audio.tobytes())

    duration = len(speaker_audio) / sample_rate

    print(f"{speaker}:")
    print(f"  segmenti : {len(pieces)}")
    print(f"  trajanje : {duration:.3f} s")
    print(f"  izlaz    : {out_path}")

    output_files.append(out_path)

print("\n=== GOTOVO ===")
print("Izrađeno glasovnih uzoraka:", len(output_files))
```

```python
import gc
import torch

from kaggle_secrets import UserSecretsClient
from kugelaudio_open import (
    KugelAudioForConditionalGenerationInference,
    KugelAudioProcessor,
)

print("=== KLONAUDIO CPU — META/LINSPACE PATCH ===")

gc.collect()

secrets = UserSecretsClient()
hf_token = secrets.get_secret("fuck_token")

# Privremeni patch:
# KlonAudio tokenizer tijekom meta-inicijalizacije poziva
# torch.linspace(...).item(), što PyTorch odbija na meta tensoru.
_original_linspace = torch.linspace

def _safe_linspace(*args, **kwargs):
    if "device" not in kwargs:
        kwargs["device"] = "cpu"
    return _original_linspace(*args, **kwargs)

torch.linspace = _safe_linspace

try:
    model = KugelAudioForConditionalGenerationInference.from_pretrained(
        "Roland-JAAI/klonaudio",
        dtype=torch.bfloat16,
        token=hf_token,
        low_cpu_mem_usage=False,
    )

    model.eval()

    processor = KugelAudioProcessor.from_pretrained(
        "Roland-JAAI/klonaudio",
        token=hf_token,
    )

    print("Model    : OK")
    print("Processor: OK")

finally:
    # Vrati originalni torch.linspace čim je učitavanje završeno
    torch.linspace = _original_linspace
```

```python
import gc
import torch

from kaggle_secrets import UserSecretsClient
from transformers.modeling_utils import no_init_weights
from kugelaudio_open import (
    KugelAudioForConditionalGenerationInference,
    KugelAudioProcessor,
)

print("=== KLONAUDIO CPU — BEZ META KONTEKSTA ===")

gc.collect()

secrets = UserSecretsClient()
hf_token = secrets.get_secret("fuck_token")

# Spremi originalnu metodu
_original_get_init_context = (
    KugelAudioForConditionalGenerationInference.get_init_context
)

# KlonAudio ne smije biti konstruiran na meta uređaju.
@classmethod
def _klonaudio_cpu_init_context(
    cls,
    is_quantized=False,
    _is_ds_init_called=False,
):
    return [no_init_weights()]

KugelAudioForConditionalGenerationInference.get_init_context = (
    _klonaudio_cpu_init_context
)

# KlonAudio je ~7B model; BF16 je važan da RAM ostane pod kontrolom.
old_dtype = torch.get_default_dtype()
torch.set_default_dtype(torch.bfloat16)

try:
    model = KugelAudioForConditionalGenerationInference.from_pretrained(
        "Roland-JAAI/klonaudio",
        dtype=torch.bfloat16,
        token=hf_token,
        local_files_only=True,
        low_cpu_mem_usage=False,
    )

    model.eval()

finally:
    torch.set_default_dtype(old_dtype)
    KugelAudioForConditionalGenerationInference.get_init_context = (
        _original_get_init_context
    )

processor = KugelAudioProcessor.from_pretrained(
    "Roland-JAAI/klonaudio",
    token=hf_token,
    local_files_only=True,
)

print("Model    : OK")
print("Processor: OK")
print("Device   :", next(model.parameters()).device)
print("Dtype    :", next(model.parameters()).dtype)
```
